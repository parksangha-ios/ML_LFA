```python
# ---------------------------------------------------------
# 라이브러리 불러오기
# ---------------------------------------------------------
import os
import numpy as np
import pandas as pd
import tensorflow as tf

from itertools import combinations
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Concatenate
from tensorflow.keras.optimizers import Adam

import matplotlib.pyplot as plt


# ---------------------------------------------------------
# GPU 메모리 증가 허용
# ---------------------------------------------------------
gpus = tf.config.experimental.list_physical_devices('GPU')
for gpu in gpus:
    tf.config.experimental.set_memory_growth(gpu, True)


# ---------------------------------------------------------
# 데이터 경로 설정
# ---------------------------------------------------------
image_data_dir = 'C:/Users/USER/Desktop/240621_256 - 복사본/'
csv_file = 'C:/Users/USER/Desktop/image_CNN W 8000-100.CSV'

IMG_WIDTH = 256
IMG_HEIGHT = 256
IMG_SIZE = (IMG_WIDTH, IMG_HEIGHT)


# ---------------------------------------------------------
# CSV 로드
# ---------------------------------------------------------
data = pd.read_csv(csv_file)

df = pd.DataFrame({
    'filename': data.iloc[:, 0].values,
    'TC': data.iloc[:, 2].values
})

df['SP'] = df['filename'].apply(lambda x: x.split('_')[0])


# ---------------------------------------------------------
# SP별 30장씩 샘플링
# ---------------------------------------------------------
TARGET_PER_SP = 30
SP_LIST = ['SP05', 'SP08', 'SP13', 'SP17']

filtered_df = []

for sp in SP_LIST:
    sp_df = df[df['SP'] == sp]

    if len(sp_df) < TARGET_PER_SP:
        raise ValueError(f"{sp} 이미지 부족")

    filtered_df.append(sp_df.sample(TARGET_PER_SP, random_state=42))

filtered_df = pd.concat(filtered_df).reset_index(drop=True)
print("총 이미지:", len(filtered_df))


# ---------------------------------------------------------
# ⭐ SP별 4장 조합 생성
# ---------------------------------------------------------
pairs = []
pair_labels = []

for sp in SP_LIST:
    sp_files = filtered_df[filtered_df['SP'] == sp]['filename'].values
    sp_tc = filtered_df[filtered_df['SP'] == sp]['TC'].values[0]

    combos = list(combinations(sp_files, 4))
    print(f"{sp} → 4장 조합 수: {len(combos)}")

    pairs.extend(combos)
    pair_labels.extend([sp_tc] * len(combos))

print("전체 조합 수:", len(pairs))


# ---------------------------------------------------------
# 이미지 경로 + label
# ---------------------------------------------------------
img1_paths = [os.path.join(image_data_dir, p[0]) for p in pairs]
img2_paths = [os.path.join(image_data_dir, p[1]) for p in pairs]
img3_paths = [os.path.join(image_data_dir, p[2]) for p in pairs]
img4_paths = [os.path.join(image_data_dir, p[3]) for p in pairs]
labels = np.array(pair_labels, dtype=np.float32)


# ---------------------------------------------------------
# TC MinMax 정규화
# ---------------------------------------------------------
y_scaler = MinMaxScaler()
labels_scaled = y_scaler.fit_transform(labels.reshape(-1, 1)).flatten()


# ---------------------------------------------------------
# Train / Validation split
# ---------------------------------------------------------
indices = np.arange(len(labels_scaled))

train_idx, val_idx = train_test_split(
    indices, test_size=0.2, random_state=42, shuffle=True
)

train_data = (
    np.array(img1_paths)[train_idx],
    np.array(img2_paths)[train_idx],
    np.array(img3_paths)[train_idx],
    np.array(img4_paths)[train_idx],
    labels_scaled[train_idx]
)

val_data = (
    np.array(img1_paths)[val_idx],
    np.array(img2_paths)[val_idx],
    np.array(img3_paths)[val_idx],
    np.array(img4_paths)[val_idx],
    labels_scaled[val_idx]
)


# ---------------------------------------------------------
# ⭐ tf.data 전처리 (4장)
# ---------------------------------------------------------
def load_4images(p1, p2, p3, p4, y):

    def load_img(path):
        img = tf.io.read_file(path)
        img = tf.image.decode_jpeg(img, channels=3)
        img = tf.image.resize(img, IMG_SIZE)
        img = tf.cast(img, tf.float32) / 255.0
        return img

    return (
        load_img(p1),
        load_img(p2),
        load_img(p3),
        load_img(p4)
    ), y


# ---------------------------------------------------------
# Dataset 구성
# ---------------------------------------------------------
BATCH_SIZE = 8   # ⭐ 4-branch라서 8 권장

train_ds = tf.data.Dataset.from_tensor_slices(train_data)
train_ds = train_ds.map(load_4images, num_parallel_calls=tf.data.AUTOTUNE)
train_ds = train_ds.shuffle(100)
train_ds = train_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

val_ds = tf.data.Dataset.from_tensor_slices(val_data)
val_ds = val_ds.map(load_4images, num_parallel_calls=tf.data.AUTOTUNE)
val_ds = val_ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)


# ---------------------------------------------------------
# Shared CNN
# ---------------------------------------------------------
def create_shared_cnn():

    inputs = Input(shape=(IMG_WIDTH, IMG_HEIGHT, 3))

    x = Conv2D(32, 3, activation='relu')(inputs)
    x = MaxPooling2D(2)(x)

    x = Conv2D(64, 3, activation='relu')(x)
    x = MaxPooling2D(2)(x)

    x = Conv2D(128, 3, activation='relu')(x)
    x = MaxPooling2D(2)(x)

    x = Conv2D(256, 3, activation='relu')(x)
    x = MaxPooling2D(2)(x)

    return Model(inputs, Flatten()(x), name="shared_cnn")


# ---------------------------------------------------------
# ⭐ 4-branch Siamese 모델
# ---------------------------------------------------------
inputs = [Input(shape=(IMG_WIDTH, IMG_HEIGHT, 3)) for _ in range(4)]
shared_cnn = create_shared_cnn()

branches = [shared_cnn(inp) for inp in inputs]
merged = Concatenate()(branches)

x = Dense(128, activation='relu')(merged)
x = Dense(64, activation='relu')(x)
output = Dense(1, activation='linear')(x)

model = Model(inputs=inputs, outputs=output)
model.compile(optimizer=Adam(), loss='mse')

model.summary()


# ---------------------------------------------------------
# 모델 학습
# ---------------------------------------------------------
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)


# ---------------------------------------------------------
# Loss 시각화
# ---------------------------------------------------------
plt.figure(figsize=(8, 6))
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='val')
plt.legend()
plt.title('4-Image Siamese Training')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


# ---------------------------------------------------------
# 1. 모델 예측 (tf.data validation dataset)
# ---------------------------------------------------------
y_pred_scaled = model.predict(val_ds).reshape(-1)

print("y_pred_scaled shape:", y_pred_scaled.shape)


# ---------------------------------------------------------
# 2. validation dataset에서 true y 수집
# ---------------------------------------------------------
y_true_scaled = np.concatenate([
    y.numpy().reshape(-1)
    for _, y in val_ds
])

print("y_true_scaled shape:", y_true_scaled.shape)


# ---------------------------------------------------------
# 3. 정규화 해제 (TC 원래 값으로 복원)
# ---------------------------------------------------------
y_pred = y_scaler.inverse_transform(
    y_pred_scaled.reshape(-1, 1)
).flatten()

y_true = y_scaler.inverse_transform(
    y_true_scaled.reshape(-1, 1)
).flatten()


# ---------------------------------------------------------
# 4. 실제 오차 계산
# ---------------------------------------------------------
errors = y_pred - y_true


# ---------------------------------------------------------
# 5. KDE (PDF) 계산
# ---------------------------------------------------------
kde = stats.gaussian_kde(errors)
x_range = np.linspace(np.min(errors), np.max(errors), 200)
pdf = kde(x_range)


# ---------------------------------------------------------
# 6. Error PDF 시각화
# ---------------------------------------------------------
plt.figure(figsize=(8, 6))
plt.hist(errors, bins=30, density=True, alpha=0.6, edgecolor='black')
plt.plot(x_range, pdf, linewidth=2)
plt.xlabel('Prediction Error (W/m·K)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.title('PDF of Prediction Error (3-Image Siamese Network)', fontsize=13)
plt.xlim(-np.max(np.abs(errors)), np.max(np.abs(errors)))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 7. 통계량 및 성능 지표 출력
# ---------------------------------------------------------
print("========= Error Statistics =========")
print(f"Mean Error     : {np.mean(errors):.4f}")
print(f"Median Error   : {np.median(errors):.4f}")
print(f"Std of Error   : {np.std(errors):.4f}")

print("\n========= Model Performance =========")
print(f"MAE  : {mean_absolute_error(y_true, y_pred):.4f}")
print(f"MSE  : {mean_squared_error(y_true, y_pred):.4f}")
print(f"RMSE : {np.sqrt(mean_squared_error(y_true, y_pred)):.4f}")
print(f"R²   : {r2_score(y_true, y_pred):.4f}")


# ---------------------------------------------------------
# 8. 실제값 vs 예측값 산점도
# ---------------------------------------------------------
plt.figure(figsize=(7, 6))
plt.scatter(y_true, y_pred, alpha=0.6)
plt.plot(
    [y_true.min(), y_true.max()],
    [y_true.min(), y_true.max()],
    linestyle='--',
    color='red'
)
plt.xlabel('True TC (W/m·K)')
plt.ylabel('Predicted TC (W/m·K)')
plt.title('True vs Predicted (Validation set - 3 Images)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

# ---------------------------------------------------------
# 설정
# ---------------------------------------------------------
image_dir = "C:/Users/USER/Desktop/predicted W(8000-100)"
output_csv = "SP_4image_group_prediction.csv"

IMG_SIZE = (256, 256)
BATCH_SIZE = 16
SP_LIST = ['SP05', 'SP08', 'SP13', 'SP17']

input_names = model.input_names   # 모델 입력 이름 자동 사용

# ---------------------------------------------------------
# SP별 이미지 수집
# ---------------------------------------------------------
sp_to_images = {sp: [] for sp in SP_LIST}

for fname in os.listdir(image_dir):
    if fname.lower().endswith((".png", ".jpg", ".jpeg", ".bmp")):
        for sp in SP_LIST:
            if fname.startswith(sp):
                sp_to_images[sp].append(os.path.join(image_dir, fname))
                break

# ---------------------------------------------------------
# SP별 4장 그룹 생성
# (남는 이미지는 버림)
# ---------------------------------------------------------
groups = []
meta_info = []

for sp, paths in sp_to_images.items():
    np.random.shuffle(paths)
    n_groups = len(paths) // 4

    print(f"{sp}: {len(paths)}장 → {n_groups}개 그룹")

    for i in range(n_groups):
        group_paths = paths[i*4:(i+1)*4]
        groups.append(group_paths)
        meta_info.append({
            "SP": sp,
            "group_idx": i,
            "img1": os.path.basename(group_paths[0]),
            "img2": os.path.basename(group_paths[1]),
            "img3": os.path.basename(group_paths[2]),
            "img4": os.path.basename(group_paths[3]),
        })

groups = np.array(groups)

print(f"\n총 예측 그룹 수: {len(groups)}")

# ---------------------------------------------------------
# tf.data 전처리 함수
# ---------------------------------------------------------
def load_image(path):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = tf.cast(img, tf.float32) / 255.0
    return img


def map_fn(paths):
    imgs = [load_image(paths[i]) for i in range(4)]
    return {
        input_names[0]: imgs[0],
        input_names[1]: imgs[1],
        input_names[2]: imgs[2],
        input_names[3]: imgs[3],
    }

# ---------------------------------------------------------
# Dataset 구성
# ---------------------------------------------------------
pred_ds = tf.data.Dataset.from_tensor_slices(groups)
pred_ds = pred_ds.map(map_fn, num_parallel_calls=tf.data.AUTOTUNE)
pred_ds = pred_ds.batch(BATCH_SIZE)
pred_ds = pred_ds.prefetch(tf.data.AUTOTUNE)

# ---------------------------------------------------------
# 모델 예측
# ---------------------------------------------------------
pred_scaled = model.predict(pred_ds, verbose=1).reshape(-1)

pred_values = y_scaler.inverse_transform(
    pred_scaled.reshape(-1, 1)
).flatten()

# ---------------------------------------------------------
# 결과 정리 및 저장
# ---------------------------------------------------------
for info, pred in zip(meta_info, pred_values):
    info["predicted_TC"] = float(pred)

results_df = pd.DataFrame(meta_info)
results_df.to_csv(output_csv, index=False, encoding="utf-8-sig")

print("Prediction finished!")
print(results_df.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error


# 예측 결과+실제값이 저장된 CSV 파일 경로
prediction_csv = 'C:/Users/USER/Desktop/Siamese/total_image_30/SP_4image_group_prediction.csv'


# 예측 결과 읽기
pred_df = pd.read_csv(prediction_csv)
print(f"데이터 shape: {pred_df.shape}")
print(pred_df.head())


# 열 이름 수정
pred_df.columns = ['SP', 'group_idx', 'img1', 'img2', 'img3', 'img4', 'predicted_TC', 'actual_TC']


# Null 값 제거
pred_df = pred_df.dropna(subset=['predicted_TC', 'actual_TC'])


# 데이터가 비어있는지 확인
if pred_df.empty:
    print("유효한 데이터가 없습니다. CSV 파일을 확인해주세요.")
else:
    # 실제값과 예측값 추출
    y_true = pred_df['actual_TC'].values
    y_pred = pred_df['predicted_TC'].values
        
    # 데이터가 충분한지 확인
    if len(y_true) < 2 or len(y_pred) < 2:
        print("데이터 샘플이 충분하지 않습니다. 최소 2개 이상의 샘플이 필요합니다.")
    
    else:
        # 모델 성능 지표 계산
        r2 = r2_score(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        mean_error = np.mean(y_pred - y_true)

        print(f"\n모델 평가 결과:")
        print(f"R²  = {r2:.4f}")
        print(f"RMSE = {rmse:.4f}")
        print(f"Mean Error = {mean_error:.4f}")


        # 예측 vs 실제값 산점도
        plt.figure(figsize=(10, 8))
        plt.scatter(y_true, y_pred, alpha=0.6)
        plt.plot([min(y_true), max(y_true)], [min(y_true), max(y_true)], 'r--', lw=2)
        plt.xlabel('Actual Thermoconductivity')
        plt.ylabel('Predicted Thermoconductivity')
        plt.title('Actual vs Predicted Thermoconductivity')
        plt.text(0.05, 0.95, f'R² = {r2:.4f}\nRMSE = {rmse:.4f}', transform=plt.gca().transAxes, 
                 verticalalignment='top', bbox=dict(facecolor='white', alpha=0.6, edgecolor='gray'))
        plt.grid(True, alpha=0.4)
        plt.tight_layout()
        plt.show()


        # 오차 분포 (히스토그램)
        errors = y_pred - y_true
        plt.figure(figsize=(10, 6))
        plt.hist(errors, bins=30, edgecolor='black')
        plt.axvline(0, color='red', linestyle='--', linewidth=2)
        plt.xlabel('Prediction Error')
        plt.ylabel('Frequency')
        plt.title('Distribution of Prediction Errors')
        plt.grid(True, alpha=0.4)
        plt.tight_layout()
        plt.show()


        # 비교 테이블 저장
        comparison_df = pred_df.copy()
        comparison_df['error'] = errors
        print("\n예측값과 실제값 비교:")
        print(comparison_df)
        
        
        # CSV 파일로 비교 결과 저장
        output_csv = 'SP_4image_group_result_comparison.csv'
        comparison_df.to_csv(output_csv, index=False, encoding='utf-8-sig')
        print(f"\n예측값 비교 결과 저장 완료: {output_csv}")
```
