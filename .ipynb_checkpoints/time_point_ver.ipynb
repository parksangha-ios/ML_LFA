```python
import numpy as np
import pandas as pd
import torch
from torch import nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error
from scipy import stats
import matplotlib.pyplot as plt

# ----------------------------------------
# 0. 공통 설정
# ----------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

seed = 42
np.random.seed(seed)
torch.manual_seed(seed)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(seed)

# ----------------------------------------
# 1. 데이터 불러오기 (Long Format CSV)
#    CSV 경로만 상황에 맞게 바꾸면 됩니다.
#    예시:
#    - 200포인트: ideal_time_200.csv
#    - 100포인트: ideal_time_100.csv
#    - 50포인트:  ideal_time_50.csv
#    - 30포인트:  ideal_time_30.csv
#    - 10포인트:  ideal_time_10.csv
# ----------------------------------------
csv_path = "C:/Users/Heum/Desktop/Sang Ha Park/CODE/ML_LFA/ideal_time_50.csv"
df = pd.read_csv(csv_path)

print("원본 Long Format shape:", df.shape)

# Expt_ID, Time_Step 순으로 정렬
df = df.sort_values(by=["Expt_ID", "Time_Step"])

# ----------------------------------------
# 2. Long Format -> 3D Tensor (N, N_TIME, 3)
#    N_TIME과 샘플 개수 자동 감지
# ----------------------------------------
# 시간 포인트 개수 (예: 200, 100, 50, 30)
N_TIME = df["Time_Step"].nunique()
# 실험(그래프) 개수
N_SAMPLES = df["Expt_ID"].nunique()

# 일관성 체크 (선택적이지만, 데이터 확인용으로 유용)
assert N_SAMPLES * N_TIME == df.shape[0], \
    f"행 수 불일치: N_SAMPLES({N_SAMPLES}) * N_TIME({N_TIME}) != df.shape[0]({df.shape[0]})"

print(f"감지된 N_TIME: {N_TIME}")
print(f"감지된 샘플(그래프) 개수: {N_SAMPLES}")

# (1) Input: V, t_log, L_log
X_V = df["V"].values.reshape(N_SAMPLES, N_TIME)
X_T = df["t_log"].values.reshape(N_SAMPLES, N_TIME)
X_L = df["L_log"].values.reshape(N_SAMPLES, N_TIME)

# (N_SAMPLES, N_TIME, 3)
X_3d = np.stack([X_V, X_T, X_L], axis=2).astype("float32")

# (2) Target: alpha_log (N_TIME개마다 같은 값이므로 한 번씩만 뽑기)
y = df["alpha_log"].values[::N_TIME].astype("float32")

print("입력 차원 (X):", X_3d.shape)  # (N_SAMPLES, N_TIME, 3)
print("타겟 차원 (y):", y.shape)     # (N_SAMPLES,)

# ----------------------------------------
# 3. Train / Test 분할 (+ validation)
# ----------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_3d, y, test_size=0.2, random_state=42
)

# 이미 0~1, log 스케일이므로 추가 스케일링 생략
X_train_scaled = X_train
X_test_scaled  = X_test

# Keras 의 validation_split=0.2와 동일하게
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_scaled, y_train, test_size=0.2, random_state=42
)

# numpy -> torch tensor
X_tr_tensor   = torch.from_numpy(X_tr)
y_tr_tensor   = torch.from_numpy(y_tr).unsqueeze(1)
X_val_tensor  = torch.from_numpy(X_val)
y_val_tensor  = torch.from_numpy(y_val).unsqueeze(1)
X_test_tensor = torch.from_numpy(X_test_scaled)
y_test_tensor = torch.from_numpy(y_test).unsqueeze(1)

batch_size = 64
train_loader = DataLoader(
    TensorDataset(X_tr_tensor, y_tr_tensor),
    batch_size=batch_size, shuffle=True
)
val_loader = DataLoader(
    TensorDataset(X_val_tensor, y_val_tensor),
    batch_size=batch_size, shuffle=False
)
test_loader = DataLoader(
    TensorDataset(X_test_tensor, y_test_tensor),
    batch_size=batch_size, shuffle=False
)

# ----------------------------------------
# 4. PyTorch MLP 모델 (N_TIME에 따라 입력 차원 자동 설정)
# ----------------------------------------
class MLPRegressor(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),                 # (N_TIME, 3) -> N_TIME * 3
            nn.Linear(input_dim, 256),    # input_dim = N_TIME * 3
            nn.ReLU(),
            nn.Linear(256, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 1)             # alpha_log 예측
        )

    def forward(self, x):
        return self.net(x)

input_dim = N_TIME * 3
print("입력 차원 (Flatten 후):", input_dim)

model = MLPRegressor(input_dim=input_dim).to(device)
print(model)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# ----------------------------------------
# 5. 학습 루프 (Early Stopping 포함)
# ----------------------------------------
def train_model(model, train_loader, val_loader, optimizer, criterion,
                n_epochs=300, patience=20):
    history = {"loss": [], "val_loss": [], "mae": [], "val_mae": []}
    best_val_loss = float("inf")
    best_state = None
    epochs_no_improve = 0

    for epoch in range(1, n_epochs + 1):
        model.train()
        train_losses = []
        train_maes = []

        for xb, yb in train_loader:
            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()
            preds = model(xb)
            loss = criterion(preds, yb)
            loss.backward()
            optimizer.step()

            train_losses.append(loss.item())
            train_maes.append(torch.mean(torch.abs(preds - yb)).item())

        train_loss = float(np.mean(train_losses))
        train_mae  = float(np.mean(train_maes))

        model.eval()
        val_losses = []
        val_maes = []
        with torch.no_grad():
            for xb, yb in val_loader:
                xb = xb.to(device)
                yb = yb.to(device)
                preds = model(xb)
                loss = criterion(preds, yb)
                val_losses.append(loss.item())
                val_maes.append(torch.mean(torch.abs(preds - yb)).item())

        val_loss = float(np.mean(val_losses))
        val_mae  = float(np.mean(val_maes))

        history["loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["mae"].append(train_mae)
        history["val_mae"].append(val_mae)

        print(
            f"Epoch {epoch:03d} | "
            f"loss={train_loss:.6f} val_loss={val_loss:.6f} "
            f"mae={train_mae:.6f} val_mae={val_mae:.6f}"
        )

        if val_loss < best_val_loss - 1e-8:
            best_val_loss = val_loss
            best_state = model.state_dict()
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f"Early stopping triggered at epoch {epoch}")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    return history

history = train_model(
    model, train_loader, val_loader,
    optimizer, criterion,
    n_epochs=300, patience=20
)

# ----------------------------------------
# 6. Best Epoch 기준 훈련 지표 테이블
# ----------------------------------------
epochs = np.arange(1, len(history["loss"]) + 1)
best_epoch = int(np.argmin(history["val_loss"]))  # 0-based index
print("Best epoch (1-based):", best_epoch + 1)

final_metrics = pd.DataFrame({
    "Metric": ["Loss (MSE)", "MAE"],
    "Training": [history["loss"][best_epoch],
                 history["mae"][best_epoch]],
    "Validation": [history["val_loss"][best_epoch],
                   history["val_mae"][best_epoch]],
})
print("\n=== Final Epoch Training Metrics (best val_loss 기준) ===")
print(final_metrics.to_string(index=False))

# ----------------------------------------
# 7. Test Set 평가 (log10 공간 + alpha 공간)
# ----------------------------------------
model.eval()
y_true_list = []
y_pred_list = []

with torch.no_grad():
    for xb, yb in test_loader:
        xb = xb.to(device)
        preds = model(xb)
        y_pred_list.append(preds.cpu().numpy())
        y_true_list.append(yb.numpy())

y_pred_log = np.concatenate(y_pred_list, axis=0).reshape(-1)
y_true_log = np.concatenate(y_true_list, axis=0).reshape(-1)

# log10(alpha) 공간
test_mse_log = mean_squared_error(y_true_log, y_pred_log)
test_mae_log = np.mean(np.abs(y_pred_log - y_true_log))
print(f"\nTest MSE (log10 space): {test_mse_log:.6f}")
print(f"Test MAE (log10 space): {test_mae_log:.6f}")

# 원래 alpha 공간
alpha_true = 10 ** y_true_log
alpha_pred = 10 ** y_pred_log

abs_err_alpha = np.abs(alpha_pred - alpha_true)
mae_alpha  = abs_err_alpha.mean()
mape_alpha = (abs_err_alpha / np.maximum(alpha_true, 1e-12)).mean() * 100.0

print(f"원래 α 공간 MAE:  {mae_alpha:.5f}")
print(f"원래 α 공간 MAPE: {mape_alpha:.2f} %")

# ----------------------------------------
# 8. 학습 커브 및 회귀 진단 플롯
# ----------------------------------------

# 8-1. Training / Validation Loss
plt.figure(figsize=(8, 6))
plt.plot(epochs, history["loss"], label="Train Loss")
plt.plot(epochs, history["val_loss"], label="Val Loss")
plt.axvline(best_epoch + 1, color="k", linestyle="--", label="Best Epoch")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Training and Validation Loss (log10 alpha)")
plt.grid(True, linestyle="--", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 8-2. True vs Predicted (log10 alpha) + R^2, RMSE
r2_log   = r2_score(y_true_log, y_pred_log)
rmse_log = np.sqrt(mean_squared_error(y_true_log, y_pred_log))

plt.figure(figsize=(8, 6))
plt.scatter(y_true_log, y_pred_log, alpha=0.5)
min_val = min(y_true_log.min(), y_pred_log.min())
max_val = max(y_true_log.max(), y_pred_log.max())
plt.plot([min_val, max_val], [min_val, max_val], "r--", lw=2, label="y = x")

plt.xlabel("True log10(alpha)")
plt.ylabel("Predicted log10(alpha)")
plt.title("True vs Predicted log10(alpha) (Test set)")
plt.text(
    0.05, 0.95,
    f"R^2 = {r2_log:.4f}\nRMSE = {rmse_log:.4f}",
    transform=plt.gca().transAxes,
    verticalalignment="top",
    fontsize=12,
    bbox=dict(boxstyle="round", facecolor="white", alpha=0.8),
)
plt.grid(True, linestyle="--", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 8-3. Error Histogram + KDE (alpha 공간)
errors_alpha = alpha_pred - alpha_true

kde_alpha = stats.gaussian_kde(errors_alpha)
x_range_alpha = np.linspace(errors_alpha.min(), errors_alpha.max(), 400)
pdf_alpha = kde_alpha(x_range_alpha)

plt.figure(figsize=(8, 6))
plt.hist(
    errors_alpha,
    bins=60,
    density=True,
    alpha=0.7,
    color="skyblue",
    edgecolor="black",
    label="Error histogram",
)
plt.plot(x_range_alpha, pdf_alpha, "r-", linewidth=2, label="KDE")

max_abs_error = max(abs(errors_alpha.min()), abs(errors_alpha.max()))
plt.xlim(-max_abs_error, max_abs_error)

plt.xlabel("Error (alpha_pred - alpha_true)")
plt.ylabel("Probability Density")
plt.title("Probability Density of Error (alpha space)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 8-4. Log Error Histogram + KDE (log10 공간)
log_errors = y_pred_log - y_true_log

kde_log = stats.gaussian_kde(log_errors)
x_range_log = np.linspace(log_errors.min(), log_errors.max(), 400)
pdf_log = kde_log(x_range_log)

plt.figure(figsize=(8, 6))
plt.hist(
    log_errors,
    bins=60,
    density=True,
    alpha=0.7,
    color="skyblue",
    edgecolor="black",
    label="Log Error histogram",
)
plt.plot(x_range_log, pdf_log, "r-", linewidth=2, label="KDE")

plt.xlabel("Log Error (pred_log10 - true_log10)")
plt.ylabel("Probability Density")
plt.title("Probability Density of Error (log10 alpha)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
