```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# -------------------------------------------------
# 1. 데이터 불러오기 (Long Format CSV)
# -------------------------------------------------
# 방금 새로 만든 Long Format CSV 경로
csv_path = "C:/Users/Heum/Desktop/Sang Ha Park/CODE/ML_LFA/lfa_dataset_v1.csv"
df = pd.read_csv(csv_path)

print("원본 Long Format shape:", df.shape)
# 예상: (100000, 6) -> 500개 그래프 * 200 스텝

# 데이터 정렬 보장 (ID순, 시간순) - 중요!
df = df.sort_values(by=["Expt_ID", "Time_Step"])

# -------------------------------------------------
# 2. 데이터 구조 변환 (Reshaping) ★ 핵심 변경 ★
#    Long Format(줄글) -> 3D Tensor(입체)
# -------------------------------------------------
N_TIME = 200
# 전체 행 개수를 시간 스텝으로 나누면 총 샘플(그래프) 개수가 나옴
N_SAMPLES = df.shape[0] // N_TIME 

print(f"감지된 샘플(그래프) 개수: {N_SAMPLES}개")

# (1) Input Data (V, t_log, L_log)
#     values로 값을 꺼낸 뒤, (샘플수, 200) 형태로 재배열합니다.
X_V = df["V"].values.reshape(N_SAMPLES, N_TIME)
X_T = df["t_log"].values.reshape(N_SAMPLES, N_TIME)
X_L = df["L_log"].values.reshape(N_SAMPLES, N_TIME)

# (2) 3차원으로 합치기 (Stacking) -> (N, 200, 3)
#     Axis 2 방향(깊이 방향)으로 합칩니다.
X_3d = np.stack([X_V, X_T, X_L], axis=2).astype("float32")

# (3) Target Data (alpha_log)
#     alpha 값은 200번 동안 똑같으므로, 200개마다 하나씩만 가져오면 됩니다.
#     [::N_TIME] 슬라이싱을 이용해 첫 번째 값만 쏙쏙 뽑습니다.
y = df["alpha_log"].values[::N_TIME].astype("float32")

print("입력 차원 (X):", X_3d.shape)  # (500, 200, 3)
print("타겟 차원 (y):", y.shape)     # (500,)

# -------------------------------------------------
# 3. Train / Test 분할
#    이제 X가 그래프 단위로 묶였으므로, 여기서 나누면
#    자동으로 ID 기준으로 깔끔하게 나뉩니다.
# -------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_3d, y, test_size=0.2, random_state=42
)

# -------------------------------------------------
# 4. 입력 스케일링
#    * 중요: 3차원 데이터(Tensor)는 StandardScaler를 바로 쓸 수 없습니다.
#    * 다행히 V(0~1), t_log, L_log로 이미 범위가 잡혀있으므로 생략합니다.
#      (이전 코드와 동일하게 처리)
# -------------------------------------------------
X_train_scaled = X_train
X_test_scaled  = X_test

# -------------------------------------------------
# 5. 모델 정의 (Flatten -> MLP)
#    데이터가 (200, 3)으로 들어오므로 Input Shape 변경
# -------------------------------------------------
input_shape = (200, 3)

model = models.Sequential([
    layers.Input(shape=input_shape),
    
    # [핵심] 3차원 데이터를 ANN이 이해할 수 있게 1줄로 쫙 폅니다.
    layers.Flatten(),
    
    # 기존 MLP 구조 유지
    layers.Dense(256, activation="relu"),
    layers.Dense(256, activation="relu"),
    layers.Dense(128, activation="relu"),
    layers.Dense(1)  # alpha_log 예측
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="mse",
    metrics=["mae"]
)

model.summary()

# -------------------------------------------------
# 6. 학습
# -------------------------------------------------
es = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled, y_train,
    validation_split=0.2,
    epochs=300,
    batch_size=64, # 데이터 개수가 줄었으니 배치는 좀 줄여도 됨
    callbacks=[es],
    verbose=2
)

# -------------------------------------------------
# 7. 평가 및 시각화 (기존 코드 유지)
# -------------------------------------------------
test_loss, test_mae = model.evaluate(X_test_scaled, y_test, verbose=0)
print(f"Test MSE (log10 space): {test_loss:.6f}")
print(f"Test MAE (log10 space): {test_mae:.6f}")

# 예측
y_pred_log = model.predict(X_test_scaled).reshape(-1)

# 원래 스케일로 변환
alpha_true = 10 ** y_test
alpha_pred = 10 ** y_pred_log

# MAE / MAPE
abs_err = np.abs(alpha_pred - alpha_true)
mae_alpha = abs_err.mean()
mape_alpha = (abs_err / np.maximum(alpha_true, 1e-12)).mean() * 100

print(f"원래 α 공간 MAE: {mae_alpha:.5f}")
print(f"원래 α 공간 MAPE: {mape_alpha:.2f} %")

# 산점도 그리기
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred_log, s=8, alpha=0.6)
min_val = min(y_test.min(), y_pred_log.min())
max_val = max(y_test.max(), y_pred_log.max())
plt.plot([min_val, max_val], [min_val, max_val], "k--", linewidth=1)

plt.xlabel("True log10(alpha)")
plt.ylabel("Pred log10(alpha)")
plt.title("log10(alpha) True vs Pred")
plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.metrics import r2_score, mean_squared_error

# =========================================================
# 0. 예측값 재계산 (log10(alpha) & alpha)
# =========================================================
# log10(alpha) 예측
y_pred_log = model.predict(X_test_scaled).flatten()
y_true_log = y_test.flatten()

# 원래 alpha 공간
alpha_true = 10 ** y_true_log
alpha_pred = 10 ** y_pred_log

# ---------------------------------------------------------
# 1) Training / Validation Loss 
# ---------------------------------------------------------
loss     = history.history["loss"]
val_loss = history.history["val_loss"]
epochs   = range(1, len(loss) + 1)

plt.figure(figsize=(10, 8))
plt.plot(epochs, loss,     'r', label='Training Loss',   linewidth=2)
plt.plot(epochs, val_loss, 'b', label='Validation Loss', linewidth=2)

plt.title('Training and Validation Loss', fontsize=14)
plt.xlabel('Epochs', fontsize=12)
plt.ylabel('MSE Loss', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 2) True vs Predicted (log10 alpha) + R^2, RMSE
# ---------------------------------------------------------
r2_log   = r2_score(y_true_log, y_pred_log)
rmse_log = np.sqrt(mean_squared_error(y_true_log, y_pred_log))

plt.figure(figsize=(10, 8))
plt.scatter(y_true_log, y_pred_log, alpha=0.5)

min_val = min(y_true_log.min(), y_pred_log.min())
max_val = max(y_true_log.max(), y_pred_log.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2, label='y = x')

plt.xlabel('True log10(alpha)', fontsize=12)
plt.ylabel('Predicted log10(alpha)', fontsize=12)
plt.title('True vs Predicted log10(alpha)', fontsize=14)

# 왼쪽 위에 R^2, RMSE 표시
plt.text(0.05, 0.95,
         f'R^2 = {r2_log:.4f}\nRMSE = {rmse_log:.4f}',
         transform=plt.gca().transAxes,
         verticalalignment='top',
         fontsize=12,
         bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 3) Error Histogram + KDE (alpha 공간, 예시 2 스타일)
# ---------------------------------------------------------
# signed error (원래 alpha 단위에서)
errors = alpha_pred - alpha_true

# KDE (gaussian_kde)
kde = stats.gaussian_kde(errors)
x_range = np.linspace(errors.min(), errors.max(), 400)
pdf = kde(x_range)

plt.figure(figsize=(10, 8))
# 히스토그램 (density=True 로 PDF와 스케일 맞추기)
plt.hist(errors,
         bins=50,
         density=True,
         alpha=0.7,
         color='skyblue',
         edgecolor='black',
         label='Error histogram')

# KDE 곡선
plt.plot(x_range, pdf, 'r-', linewidth=2, label='KDE')

# x축을 대칭 범위로 (예시 2 스타일)
max_abs_error = max(abs(errors.min()), abs(errors.max()))
plt.xlim(-max_abs_error, max_abs_error)

plt.xlabel('Error (alpha_pred - alpha_true)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.title('Probability Density of Error (alpha space)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

# 기본 통계도 같이 출력
print(f"Mean Error (alpha):   {np.mean(errors):.4f}")
print(f"Median Error (alpha): {np.median(errors):.4f}")
print(f"Std of Error (alpha): {np.std(errors):.4f}")
```

```python
log_errors = y_pred_log - y_true_log

plt.figure(figsize=(10, 8))
plt.hist(log_errors, bins=60, density=True, alpha=0.7,
         color='skyblue', edgecolor='black', label='Log Error histogram')

kde = stats.gaussian_kde(log_errors)
xr = np.linspace(log_errors.min(), log_errors.max(), 400)
plt.plot(xr, kde(xr), 'r-', linewidth=2, label='KDE')

plt.xlabel('Log Error (pred_log10 - true_log10)')
plt.ylabel('Probability Density')
plt.title('Probability Density of Error (log10 alpha)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
