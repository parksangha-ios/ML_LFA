```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 방금 만든 데이터 로드
csv_path = "C:/Users/Heum/Desktop/Sang Ha Park/CODE/ML_LFA/lfa_realistic_fixed_time_100pt.csv"
df = pd.read_csv(csv_path)

# 중복된 그래프 데이터 제거 (그래프 1개당 1개의 점만 찍기 위해)
# Expt_ID별로 첫 번째 행만 가져옴
df_unique = df.drop_duplicates(subset=["Expt_ID"])

print(f"총 그래프 개수: {len(df_unique)}개")

# 2. 시각화 (Alpha Log vs L Log 분포)
plt.figure(figsize=(10, 8))
# 산점도 그리기
plt.scatter(df_unique["alpha_log"], df_unique["L_log"], 
            alpha=0.3, s=10, c='blue', edgecolors='none', label='Generated Data')

plt.title("Data Distribution check: Alpha vs L (Log Scale)", fontsize=15)
plt.xlabel("Log10 (Thermal Diffusivity)", fontsize=12)
plt.ylabel("Log10 (Thickness)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# 범위 박스 표시 (Grid가 잘 찼는지 확인)
plt.axvline(df_unique["alpha_log"].min(), color='r', linestyle='--')
plt.axvline(df_unique["alpha_log"].max(), color='r', linestyle='--')
plt.axhline(df_unique["L_log"].min(), color='r', linestyle='--')
plt.axhline(df_unique["L_log"].max(), color='r', linestyle='--')

plt.legend()
plt.show()

# 3. 히스토그램 (각 변수가 얼마나 평평한지 확인)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Alpha 분포
sns.histplot(df_unique["alpha_log"], bins=30, kde=False, ax=axes[0], color='skyblue')
axes[0].set_title("Distribution of Log(Alpha)")
axes[0].set_xlabel("Log10 Alpha")
axes[0].set_ylabel("Count")

# L 분포
sns.histplot(df_unique["L_log"], bins=20, kde=False, ax=axes[1], color='salmon')
axes[1].set_title("Distribution of Log(L)")
axes[1].set_xlabel("Log10 L")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 중복 제거: 실험(Expt_ID) 하나당 첫 번째 행만 남김
df_params = df.drop_duplicates(subset=["Expt_ID"])

plt.figure(figsize=(12, 5))

# 1. Alpha 분포 (Log Scale)
plt.subplot(1, 2, 1)
# bins를 생성한 Grid 개수와 비슷하거나 작게 잡으면 평평해 보임
sns.histplot(df_params["alpha_log"], bins=30, kde=True, color="skyblue")
plt.title("Distribution of Alpha (Log Scale) - Unique Graphs")
plt.xlabel("Log Alpha")
plt.ylabel("Count (Number of Graphs)")

# 2. L 분포 (Log Scale)
plt.subplot(1, 2, 2)
sns.histplot(df_params["L_log"], bins=20, kde=True, color="salmon")
plt.title("Distribution of Thickness (Log Scale) - Unique Graphs")
plt.xlabel("Log Thickness")

plt.tight_layout()
plt.show()
```
