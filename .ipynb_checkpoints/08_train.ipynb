```python
dataset_path = 'datasets/dbnet-2018-preprocessed'
```

```python
train_dataset_filename = "dbnet_train_dataset.csv"
test_dataset_filename = "dbnet_test_dataset.csv"
```

```python
df_train = pd.read_csv(train_dataset_filename)
df_test = pd.read_csv(test_dataset_filename)
```

```python
# define image transforms
image_transforms_train = torchvision.transforms.Compose([
    # center crop to 1080x660
    torchvision.transforms.CenterCrop((660, 1080)),
    # resize to 220x66
    torchvision.transforms.Resize((66, 220)),
    # random horizontal flip
    # torchvision.transforms.RandomHorizontalFlip(),
    # color jitter
    torchvision.transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1),
    # to tensor
    torchvision.transforms.ToTensor(),
])

image_transforms_test = torchvision.transforms.Compose([
    # center crop to 1080x660
    torchvision.transforms.CenterCrop((660, 1080)),
    # resize to 220x66
    torchvision.transforms.Resize((66, 220)),
    # to tensor
    torchvision.transforms.ToTensor(),
])
```

```python
# load datasets
from utils import DBNetDatasetIO
dataset_train = DBNetDatasetIO(csv_file=train_dataset_filename, image_transform=image_transforms_train, 
                               normalize_output=True, max_wheel_angle=100, min_wheel_angle=-100, max_vehicle_speed=60, min_vehicle_speed=0)
dataset_test = DBNetDatasetIO(csv_file=test_dataset_filename, image_transform=image_transforms_test,
                                 normalize_output=True, max_wheel_angle=100, min_wheel_angle=-100, max_vehicle_speed=60, min_vehicle_speed=0)
```

```python
from torch import nn

class DBNet(torch.nn.Module):
    def __init__(self):
        super(DBNet, self).__init__()
        from torchvision.models import ResNet18_Weights
        self.backbone = torchvision.models.resnet18(weights=ResNet18_Weights.IMAGENET1K_V1)
        # remove last fully connected layer
        self.backbone = torch.nn.Sequential(*list(self.backbone.children())[:-1])

        self.wheel_angle_head = nn.Sequential(
            torch.nn.Linear(512, 128),
            torch.nn.ReLU(),
            torch.nn.Linear(128, 1)
        )

        self.vechicle_speed_head = nn.Sequential(
            torch.nn.Linear(512, 128),
            torch.nn.ReLU(),
            torch.nn.Linear(128, 1)
        )
        
    def forward(self, x):
        x = self.backbone(x)
        x = torch.flatten(x, 1)  # Flatten the output of the backbone
        wheel_angle = self.wheel_angle_head(x)
        vehicle_speed = self.vechicle_speed_head(x)
        return wheel_angle, vehicle_speed
```

```python
model = DBNet()
```

```python
# hyperparameters
learning_rate = 0.01
batch_size = 32*8*4
num_epochs = 100

# optimizer
optimizer = torch.optim.AdamW(model.parameters(), lr=learning_rate)
```

```python
# loss function: mse for both wheel angle and vehicle speed
def criterion(predictions, targets):
    wheel_angle_pred, vehicle_speed_pred = predictions
    wheel_angle_target, vehicle_speed_target = targets
    loss_wheel_angle = nn.MSELoss()(wheel_angle_pred, wheel_angle_target)
    loss_vehicle_speed = nn.MSELoss()(vehicle_speed_pred, vehicle_speed_target)
    return loss_wheel_angle, loss_vehicle_speed
```

```python
# prepare to train
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

train_dataloader = torch.utils.data.DataLoader(dataset_train, batch_size=batch_size, shuffle=True)
test_dataloader = torch.utils.data.DataLoader(dataset_test, batch_size=batch_size, shuffle=False)
```

```python
# logger
class AverageMeter():
    def __init__(self):
        self.reset()

    def reset(self):
        self.val = 0
        self.avg = 0
        self.sum = 0
        self.count = 0

    def update(self, val, n=1):
        self.val = val
        self.sum += val * n
        self.count += n
        self.avg = self.sum / self.count
```

```python
# training function 
def spin_once(model, dataloader, optimizer, criterion, device, flag_train=True):
    if flag_train:
        model.train()
    else:
        model.eval()

    total_loss = 0.0
    total_wheel_angle_loss = 0.0
    total_vehicle_speed_loss = 0.0

    for batch_idx, (images, wheel_angle_targets, vehicle_speeds_targets) in enumerate(dataloader):
        images = images.to(device)
        wheel_angle_target = wheel_angle_targets.to(device)
        vehicle_speed_target = vehicle_speeds_targets.to(device)

        optimizer.zero_grad()
        wheel_angle_pred, vehicle_speed_pred = model(images)
        wheel_angle_loss, vehicle_speed_loss = criterion((wheel_angle_pred, vehicle_speed_pred), (wheel_angle_target, vehicle_speed_target))
        loss = wheel_angle_loss + vehicle_speed_loss

        # print loss in every 10 batches
        if batch_idx % 10 == 0:
            print(f'Batch {batch_idx}/{len(dataloader)} - Total Loss: {loss.item():.4f} (Wheel Angle Loss: {wheel_angle_loss.item():.4f}, Vehicle Speed Loss: {vehicle_speed_loss.item():.4f})')
        
        if flag_train:
            loss.backward()
            optimizer.step()

        total_loss += loss.item() * images.size(0)
        total_wheel_angle_loss += wheel_angle_loss.item() * images.size(0)
        total_vehicle_speed_loss += vehicle_speed_loss.item() * images.size(0)

    avg_loss = total_loss / len(dataloader.dataset)
    avg_wheel_angle_loss = total_wheel_angle_loss / len(dataloader.dataset)
    avg_vehicle_speed_loss = total_vehicle_speed_loss / len(dataloader.dataset)

    return avg_loss, avg_wheel_angle_loss, avg_vehicle_speed_loss
```

```python
# training loop
train_loss_meter = AverageMeter()
test_loss_meter = AverageMeter()
train_wheel_angle_loss_meter = AverageMeter()
train_vehicle_speed_loss_meter = AverageMeter()
test_wheel_angle_loss_meter = AverageMeter()
test_vehicle_speed_loss_meter = AverageMeter()

train_loss_meter.reset()
test_loss_meter.reset()
train_wheel_angle_loss_meter.reset()
train_vehicle_speed_loss_meter.reset()
test_wheel_angle_loss_meter.reset()
test_vehicle_speed_loss_meter.reset()

best_test_loss = float('inf')

for epoch in range(num_epochs):

    # write log file
    if epoch == 0:
        # if file exists, remove it
        if os.path.exists('training_log.csv'):
            print("Removing old training_log.csv")
            os.remove('training_log.csv')
        # make new file with header
        with open('training_log.csv', 'w') as f:
            print("Creating new training_log.csv with header")
            f.write('epoch,train_loss,train_wheel_angle_loss,train_vehicle_speed_loss,test_loss,test_wheel_angle_loss,test_vehicle_speed_loss\n')
    
    train_loss, train_wheel_angle_loss, train_vehicle_speed_loss = spin_once(model, train_dataloader, optimizer, criterion, device, flag_train=True)
    test_loss, test_wheel_angle_loss, test_vehicle_speed_loss = spin_once(model, test_dataloader, optimizer, criterion, device, flag_train=False)
    
    train_loss_meter.update(train_loss)
    test_loss_meter.update(test_loss)
    train_wheel_angle_loss_meter.update(train_wheel_angle_loss)
    train_vehicle_speed_loss_meter.update(train_vehicle_speed_loss)
    test_wheel_angle_loss_meter.update(test_wheel_angle_loss)
    test_vehicle_speed_loss_meter.update(test_vehicle_speed_loss)

   # save best model
    if test_loss_meter.avg < best_test_loss:
        best_test_loss = test_loss_meter.avg
        torch.save(model.state_dict(), 'best_dbnet_model.pth')
        print(f"Best model saved at epoch {epoch+1} with test loss {best_test_loss:.4f}")
        
    with open('training_log.csv', 'a') as f:
        f.write(f"{epoch+1},{train_loss_meter.avg},{train_wheel_angle_loss_meter.avg},{train_vehicle_speed_loss_meter.avg},{test_loss_meter.avg},{test_wheel_angle_loss_meter.avg},{test_vehicle_speed_loss_meter.avg}\n")
        
    print(f"Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss_meter.avg:.4f} (Wheel Angle: {train_wheel_angle_loss_meter.avg:.4f}, Vehicle Speed: {train_vehicle_speed_loss_meter.avg:.4f}), Test Loss: {test_loss_meter.avg:.4f} (Wheel Angle: {test_wheel_angle_loss_meter.avg:.4f}, Vehicle Speed: {test_vehicle_speed_loss_meter.avg:.4f})")
```

```python
# load training log
df_log = pd.read_csv('training_log.csv')
```
