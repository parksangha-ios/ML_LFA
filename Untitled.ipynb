```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. 데이터 로드
df = pd.read_csv("C:/Users/Heum/Desktop/Sang Ha Park/CODE/ML_LFA/parker_dataset_tgrid_alpha0_01_to_2000.csv")

# 2. 입력 컬럼 설정
V_cols = [col for col in df.columns if col.startswith("V_")]
X = df[V_cols + ["L_mm"]].values.astype("float32")

# 3. 타깃: log10(alpha)
alpha = df["alpha_mm2s"].values.astype("float32")
alpha_log = np.log10(alpha)   # 핵심

y = alpha_log.reshape(-1, 1)

# 4. Train / Val / Test 분할
X_train, X_tmp, y_train, y_tmp = train_test_split(
    X, y, test_size=0.3, random_state=42, shuffle=True
)
X_val, X_test, y_val, y_test = train_test_split(
    X_tmp, y_tmp, test_size=0.5, random_state=42, shuffle=True
)

# 5. 입력 스케일링
scaler_X = StandardScaler()
X_train_scaled = scaler_X.fit_transform(X_train)
X_val_scaled = scaler_X.transform(X_val)
X_test_scaled = scaler_X.transform(X_test)

# 6. 모델 정의 (구조는 일단 그대로 두되, 필요하면 줄여도 됨)
input_dim = X_train_scaled.shape[1]

model = models.Sequential([
    layers.Input(shape=(input_dim,)),
    layers.Dense(512, activation="relu"),
    layers.Dense(256, activation="relu"),
    layers.Dense(128, activation="relu"),
    layers.Dense(1, activation="linear")  # log10(α) 회귀
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="mse",
    metrics=["mae"]
)

callback_es = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled, y_train,
    validation_data=(X_val_scaled, y_val),
    epochs=300,
    batch_size=64,
    callbacks=[callback_es],
    verbose=1
)

# 7. log 공간에서의 성능
test_loss, test_mae = model.evaluate(X_test_scaled, y_test, verbose=0)
print("Test MSE (log10 space):", test_loss)
print("Test MAE (log10 space):", test_mae)

# 8. 실제 α 공간으로 복원해서 평가
y_test_log = y_test.flatten()
y_pred_log = model.predict(X_test_scaled).flatten()

y_test_real = 10 ** y_test_log
y_pred_real = 10 ** y_pred_log

abs_err = np.abs(y_pred_real - y_test_real)
rel_err = abs_err / (y_test_real + 1e-8)   # 0 근처 방어용 eps

print("원래 α 공간 MAE:", abs_err.mean())
print("원래 α 공간 MAPE(평균 상대오차):", (rel_err * 100).mean(), "%")

print("예측 α (앞 몇개):", y_pred_real[:5])
print("실제 α (앞 몇개):", y_test_real[:5])
```

```python
bins = [0, 1, 10, 100, 10000]
labels = ["<1", "1-10", "10-100", "100+"]

import pandas as pd
df_eval = pd.DataFrame({
    "alpha_true": y_test_real,
    "alpha_pred": y_pred_real
})
df_eval["bin"] = pd.cut(df_eval["alpha_true"], bins=bins, labels=labels, include_lowest=True)

for label in labels:
    sub = df_eval[df_eval["bin"] == label]
    if len(sub) == 0:
        continue
    ae = (sub["alpha_pred"] - sub["alpha_true"]).abs()
    re = ae / (sub["alpha_true"] + 1e-8)
    print(f"구간 {label}: n={len(sub)}, MAE={ae.mean():.3f}, MAPE={re.mean()*100:.1f}%")
```

```python
import matplotlib.pyplot as plt
plt.scatter(y_test_log, y_pred_log, s=5)
plt.plot([y_test_log.min(), y_test_log.max()],
         [y_test_log.min(), y_test_log.max()])
plt.xlabel("True log10(alpha)")
plt.ylabel("Pred log10(alpha)")
plt.title("log10(alpha) 예측 산점도")
plt.show()
```
